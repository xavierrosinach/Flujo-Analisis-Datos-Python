```python
import pandas as pd
import os
import markdown
from tabulate import tabulate
from datetime import datetime
import textwrap
```

```python
# Aplicamos un estilo predefinido a nuestro codigo html
def apply_style(html: str) -> str:

    return f"""
        <!DOCTYPE html>
        <html>
        <head>
        <meta charset="utf-8">
        <title>Reporte Ejecutivo</title>
        <link href="https://fonts.googleapis.com/css2?family=Roboto:wght@400;700&display=swap" rel="stylesheet">
        <style>
            body {{
                font-family: 'Roboto', sans-serif;
                background-color: #f7f9fc;
                margin: 40px;
                color: #333;
            }}
            h1, h2, h3 {{
                color: #2c3e50;
            }}
            h1 {{
                text-align: center;
                margin-bottom: 20px;
            }}
            h4 {{
                text-align: center;
            }}
            table {{
                border-collapse: collapse;
                width: 70%;
                margin: 20px 0;
                box-shadow: 0 2px 5px rgba(0,0,0,0.1);
            }}
            th, td {{
                border: 1px solid #ddd;
                padding: 12px 15px;
                text-align: center;
            }}
            th {{
                background-color: #4CAF50;
                color: white;
                font-weight: bold;
            }}
            tr:nth-child(even) {{
                background-color: #f2f2f2;
            }}
            tr:hover {{
                background-color: #d1e7dd;
            }}
            footer {{
                margin-top: 40px;
                text-align: center;
                font-size: 0.9em;
                color: #555;
            }}
            footer a {{
                text-decoration: none;
                font-weight: bold;
            }}
            img {{
                display: block;
                margin-left: auto;
                margin-right: auto;
                width: 50%;
            }}
        </style>
        </head>
        <body>
        {html}
        <footer>
            <b>Xavier Rosinach Capell</b><br>
            <i>Sports Data Scientist and Engineer</i><br><br>
            <a href="https://github.com" target="_blank" style="color:#333; margin-right:15px;">GitHub</a>
            <a href="https://www.linkedin.com" target="_blank" style="color:#0077b5;">LinkedIn</a><br><br>
            <span style="font-size:0.8em; color:#999;">&copy; {datetime.now().year} Xavier Rosinach Capell. Todos los derechos reservados.</span>
        </footer>
        
        </body>
        </html>
        """

# Transformamos en dataframe a un formato idonio para subirlo al HTML
def add_df(df_path: str):

    if os.path.exists(df_path):
        df = pd.read_csv(df_path, encoding='utf-8')
        return tabulate(df, headers='keys', tablefmt='github', showindex=False)
    return ""
```

```python
# Creación del report de una temporada
def season_report_creator(figures_path: str, season: str, reports_path: str = ".") -> None:

    # Cabecera con títulos
    markdown_content = textwrap.dedent(f"""
    # *REPORT* DE LA TEMPORADA {season}
    #### Comparación de estadísticas entre ligas
    ---
    """)

    # Añadimos dataframe con información general
    df_md = add_df(os.path.join(figures_path, 'df.csv'))
    markdown_content += textwrap.dedent(f"""
    ## 1. Estadísticas generales

    {df_md}

    La siguiente tabla nos muestra las estadísticas principales de diferentes ligas. Estas métricas representan:

    - **Attack Value**: Suma de la media de los goles esperados, grandes ocasiones creadas, y *attack value* por 90 minutos de todos los jugadores de aquella liga. Es un valor que nos da medida a la **ofensividad** de la liga.
    - **Defense Value**: Suma de la media de acciones defensivas, duelos ganados, duelos aéreos y intercepciones por 90 minutos de los jugadores de la liga. Mide la **eficacia defensiva** de una liga.
    - **Progression Value**: Suma de la media de prograsiones y pases normalizados por 90 minutos por jugador de la liga. Nos ayuda a ver la **verticalidad** de las ligas.
    ---
    """)

    # Agregar visualizacion 1
    grafico_path = os.path.join(figures_path, "grafico.png")
    if os.path.exists(grafico_path):
        markdown_content += textwrap.dedent(f"""
        ## {sectioner}. Gráfico de estadísticas

        ![Gráfico de estadísticas]({grafico_path})
        ---
        """)
        sectioner += 1

    # Guardado MD
    md_file = os.path.join(reports_path, f"reporte.md")
    with open(md_file, "w", encoding="utf-8") as f:
        f.write(markdown_content)

    # Conversion HTML
    html_content = markdown.markdown(markdown_content, extensions=["tables"])
    html_file = os.path.join(reports_path, f"reporte.html")
    with open(html_file, "w", encoding="utf-8") as f:
        f.write(apply_style(html_content))

    # Borrar documento MD
    os.remove(md_file)
```

```python
season_report_creator('a', '24/25')
```
